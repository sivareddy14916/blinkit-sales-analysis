```python
df["Item Fat Content"]=df["Item Fat Content"].replace({
                                                       'reg':'Regular',
                                                       'low fat':'Low Fat',
                                                       'LF':'Low Fat'
                                                       })
```

# **KPIs Requirements**

```python
#Total Sales
Total_Sales=df['Sales'].sum()

#Avg Sales
Avg_Sales=df['Sales'].mean()

#Number of Items
Number_of_Items=df['Item Identifier'].count()

#Avg Rating
Avg_Rating=df['Rating'].mean()

print(f"Total Sales: ${Total_Sales:,.1f} | Avg Sales: ${Avg_Sales:,.1f} | Total Items: {Number_of_Items} | Avg Rating: {Avg_Rating:,.1f}")
```

# **Charts Requirements**

**Total Sales by Fat Content**

```python
sales_by_fat=df.groupby('Item Fat Content')['Sales'].sum()
print(sales_by_fat)
```

**Pie chart**

```python
plt.pie(sales_by_fat,autopct='%.1f%%',labels = sales_by_fat.index,explode=(0.1,0),
                                     colors=['yellowgreen', 'lightskyblue'],shadow=True,startangle=90)
plt.title('Sales by Fat Content')
plt.axis('equal')
plt.show()
```

**Total Sales by Item Type**

```python
sales_by_item_type=df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)
print(sales_by_item_type)
```

**Bar chart**

```python
plt.figure(figsize=(10,6))
bars=plt.bar(sales_by_item_type.index,sales_by_item_type.values)
plt.title('Sales by Item Type')
plt.xlabel('Item Type')
plt.ylabel('Total Sales')
plt.xticks(rotation=-90)
for bar in bars:
  plt.text(
      bar.get_x() + bar.get_width() /  2, bar.get_height(),
      f'{bar.get_height():,.0f}',ha='center',va='bottom',fontsize =8)
plt.tight_layout()
plt.show()
```

**Fat Content by Outlet for Total Sales**

```python
fat_outlet_by_sales=df.groupby(['Outlet Location Type','Item Fat Content'])['Sales'].sum().unstack()
print(fat_outlet_by_sales)
```

```python
fat_outlet_by_sales.plot(kind='bar')
plt.title('Outlet tier by item fat content')
plt.xlabel('outlet location tier')
plt.ylabel('total sales')
plt.show()
```

**Total Sales By Outlet Establishment Year**

```python
outlet_year_sales=df.groupby('Outlet Establishment Year')['Sales'].sum()
plt.figure(figsize=(10,5))
plt.plot(outlet_year_sales.index,outlet_year_sales.values,marker='o',linestyle='-')
for x,y in zip(outlet_year_sales.index,outlet_year_sales.values):
  plt.text(x,y,f'{y:,.0f}',ha='center',va='bottom')
plt.show()
```

**Total Sales by Outlet Size**

```python
sales_by_outlet_size=df.groupby('Outlet Size')['Sales'].sum()
sales_by_outlet_size
```

```python
plt.figure(figsize=(4,4))
plt.pie(sales_by_outlet_size,labels=sales_by_outlet_size.index,autopct='%1.1f%%',startangle=90,shadow=True)
plt.title('Sales by Outlet Size')
plt.tight_layout()
plt.show()
```

**Sales By Outlet Location**

```python
sales_by_outlet_location=df.groupby('Outlet Location Type')['Sales'].sum().reset_index()
sales_by_outlet_location
ax=sns.barplot(x='Sales',y='Outlet Location Type',data=sales_by_outlet_location)
plt.show()
```
